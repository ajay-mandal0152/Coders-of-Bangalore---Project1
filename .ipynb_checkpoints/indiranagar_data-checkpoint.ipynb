```python
with open("initialdata.txt", encoding='utf-8') as f:
    data = f.read()
```

```python
data
```

```python
chunks = data.split("\n\n") 
chunks = [c for c in chunks if len(c)>3]
chunks
```

```python
def parse_chunks(chunk):
    chunk = chunk.strip()
    sep_chunk = chunk.split('\n')
    user_name = sep_chunk[0]
    no_of_posts = sep_chunk[1]
    no_of_followers = sep_chunk[2]
    no_of_followings = sep_chunk[3]
    name = sep_chunk[4]
    type_of_page = sep_chunk[5]
    bio = "\n".join(sep_chunk[6:])
    print(user_name, no_of_posts, no_of_followers, no_of_followings, name, type_of_page, bio, sep="\n")
parse_chunks(chunks[0])
```
